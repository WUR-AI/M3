# ReAct baseline on Cybench

The Ireland maize folder is one multi-CSV dataset, the same way M³ runs it. All nine CSVs are registered together, and the model uses the multi-CSV tools. The CSV text is not placed in the prompt. When the model stops calling tools, the next step writes `croissant_standard_subset`.

Results go to `outputs/react_cybench_<model>/cybench_IE_react.json`. An existing result file is kept and not replaced.

Run the setup cell, then the run cell. After a kernel restart, rerun the setup cell first.

```python
import os
import sys
from pathlib import Path

from dotenv import load_dotenv

BASE = next(
    (path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'src/config.py').is_file()),
    None,
)
if BASE is None:
    raise FileNotFoundError('Could not locate the metadata_agent project root')
sys.path.insert(0, str(BASE))
load_dotenv(BASE / '.env')
os.environ['LLM_PROVIDER'] = 'surf'

from src.baseline.ReAct import CONTEXT_WINDOWS, DEFAULT_MODELS, run_cybench

OUTPUTS_ROOT = BASE / 'outputs'
PROCESS_TIMEOUT_SECONDS = 600
CYBENCH_DIR = BASE / 'data/experiment/cybench-maize-IE/IE'
STANDARD_NAME = 'croissant_standard_subset'
EXPECTED_FILES = 9

if not os.getenv('SURF_API_KEY'):
    raise ValueError('SURF_API_KEY is missing from metadata_agent/.env')
if not os.getenv('SURF_API_BASE'):
    raise ValueError('SURF_API_BASE is missing from metadata_agent/.env')
csv_files = sorted(CYBENCH_DIR.glob('*.csv'))
if len(csv_files) != EXPECTED_FILES:
    raise ValueError(f'Expected {EXPECTED_FILES} Cybench CSV files, found {len(csv_files)}')

print(f'Project: {BASE}')
print(f'Cybench: {CYBENCH_DIR}')
print(f'CSV files: {len(csv_files)}')
for path in csv_files:
    print(f'  {path.name}')
print(f'Standard: {STANDARD_NAME}')
print(f'Models: {len(DEFAULT_MODELS)}')
for model in DEFAULT_MODELS:
    print(f'  {model}: context window {CONTEXT_WINDOWS[model]}')
```

## Run Llama

One multi-CSV ReAct run with `default-text-large`. Restarting this cell skips the run when `cybench_IE_react.json` or `cybench_IE_react.skipped.json` already exists.

```python
summaries = run_cybench(
    data_dir=CYBENCH_DIR,
    outputs_root=OUTPUTS_ROOT,
    standard_name=STANDARD_NAME,
    models=['default-text-large'],
    timeout_seconds=PROCESS_TIMEOUT_SECONDS,
)
summaries
```
